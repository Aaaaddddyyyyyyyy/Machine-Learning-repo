```python
from sklearn import datasets
# The load_boston() function has been removed from scikit-learn due to ethical concerns with the dataset.
# As an alternative, we will use the California housing dataset.
housing = datasets.fetch_california_housing()
X_housing, y_housing = housing.data, housing.target
print('Dataset feature names: ' + str(housing.feature_names))
print('Dataset feature size: ' + str(housing.data.shape))
print('Dataset target size: ' + str(housing.target.shape))
```

```python
from sklearn.metrics import r2_score
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import GridSearchCV
from sklearn.tree import DecisionTreeRegressor # Changed to Regressor
from sklearn.neighbors import KNeighborsRegressor # Added Regressor
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train,X_test,y_train,y_test=train_test_split(X_housing,y_housing,test_size=0.2,random_state=43)
print('train/test size :',X_train.shape,X_test.shape,y_train.shape,y_test.shape)
```

```python
lr= LinearRegression()
kn=KNeighborsRegressor() # Changed to Regressor
dt=DecisionTreeRegressor() # Changed to Regressor
```

```python
lr.fit(X_train,y_train)
dt.fit(X_train,y_train)
kn.fit(X_train,y_train)
```

```python
y_pred1=lr.predict(X_test)
y_pred2=kn.predict(X_test)
y_pred3=dt.predict(X_test)
```

```python
print('r2 score of lr',r2_score(y_test,y_pred1))
print('r2 score of kn', r2_score(y_test,y_pred2))
print('r2 score of dt',r2_score(y_test,y_pred3))
```

```python
from sklearn.ensemble import BaggingRegressor
bag_regressor=BaggingRegressor(random_state=2)
bag_regressor.fit(X_train,y_train)
```

```python
y_preds= bag_regressor.predict(X_test)
print('trainng coeeficient of r2 score :',bag_regressor.score(X_train,y_train))
print('test coeefcientr of r2 score :',bag_regressor.score(X_test,y_test))
```
