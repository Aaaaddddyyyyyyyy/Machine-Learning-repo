```python
import pandas as pd
import numpy as np
```

```python
df=pd.read_csv('/content/heart.csv')
```

```python
## making the prediction from the model
y_pred1= clf1.predict(X_test)
y_pred2=clf2.predict(X_test)
```

```python
from sklearn.metrics import accuracy_score,confusion_matrix
print('Accuarc of logistics regression :',accuracy_score(y_test,y_pred1))
print('Accuaracy of decisdion tree :',accuracy_score(y_test,y_pred2))
```
