```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
df=pd.read_csv('/content/placement1.csv')
```

```python
df.head()
```

```python
plt.scatter(df['cgpa'],df['package'])
plt.xlabel("CGPA")
plt.ylabel('PACKAGE')
plt.show()
```

```python
X=df.iloc[:,0:1]          # separating X (input column) and y(output column)
y=df.iloc[:,-1]
X.head()
```

```python
# Next step  to train a ml model
from sklearn.model_selection import train_test_split          # spliting data into training and test
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.2,random_state=2)
```

```python
## Applying the LInear regression
from sklearn.linear_model import LinearRegression
```

```python
# making a class
lr=LinearRegression()
```

```python
# Using fit class to train the model
lr.fit(X_train,y_train)
```
