```python
from sklearn.datasets import make_regression
import pandas as pd
import numpy as np
import plotly.express as  px
import plotly.graph_objects as go

from sklearn.metrics import  mean_absolute_error,mean_squared_error,r2_score
```

```python
X,y=make_regression(n_samples=100,n_features=2, n_informative=2,n_targets=1, noise=8)
```

```python
df=pd.DataFrame({'feature1':X[:,0],'feature2':X[:,1],'target':y})
```

```python
df.head()
```

```python
fig = px.scatter_3d(df,x='feature1',y='feature2',z='target')
fig.show()
```

```python
# training  the model

from sklearn.model_selection  import train_test_split
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.2,random_state=2)
```

```python
from sklearn.linear_model import LinearRegression
lr=LinearRegression()

```

```python
lr.fit(X_train,y_train)
```

```python
y_pred=lr.predict(X_test)
```

```python
print('MAE',mean_absolute_error(y_test,y_pred))
print('MSE',mean_squared_error(y_test,y_pred))
print('R2 score',r2_score(y_test,y_pred))
```

```python
lr.coef_        # they are the value sof beta1,beta2
```

```python
lr.intercept_     # intercept
```

Polynomial regression

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import PolynomialFeatures,StandardScaler
from sklearn.linear_model import LinearRegression,SGDRegressor
from sklearn.metrics import r2_score
from sklearn.pipeline import Pipeline
```

```python
X = 6*np.random.rand(200,1)-3                 # y = 0.8x*2+ 0.9x+ 2
y = 0.8*X**2+ 0.9*X+2+ np.random.randn(200,1)
```

```python
plt.plot(X,y,'g')
plt.xlabel('X')
plt.ylabel('y')
plt.show()
```

```python
## split data into train test
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.2,random_state=2)
```

```python
## create a class object
lr=LinearRegression()
```

```python
lr.fit(X_train,y_train)       # apply7ing the linear regression algo
```

```python
## calculate the r2 score
y_pred=lr.predict(X_test)
r2_score(y_test,y_pred)
```

```python
plt.plot(X_train,lr.predict(X_train),color='red')
plt.plot(X,y,'y')
plt.xlabel('X')
plt.ylabel('Y')
plt.show()
```

```python
## Applying the polynomial linear regression
poly=PolynomialFeatures(degree=2,include_bias=True)       ### making class of polynomialfeatutes  in object poly

X_train_trans=poly.fit_transform(X_train)       ## include bias used to add or remove the x^0 values
X_test_trans=poly.transform(X_test)
```

```python
print(X_train[0])           ## polyomial always apply on inputy not on output y
print(X_train_trans[0])
```

```python
lr=LinearRegression()
lr.fit(X_train_trans,y_train)
```

```python
y_pred= lr.predict(X_test_trans)
r2_score(y_test,y_pred)
```

```python
print(lr.coef_)
print(lr.intercept_)
```
