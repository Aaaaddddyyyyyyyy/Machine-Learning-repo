```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
from sklearn.datasets import make_classification      # make classification to make classification datasets
X,y=make_classification(n_samples=100,n_informative=1,n_features=2,n_redundant=0,n_classes=2,n_clusters_per_class=1,random_state=1,class_sep=10)
```

```python
np.insert(X,0,1,axis=1)
```

```python
plt.figure(figsize=(10,6))
plt.scatter(X[:,0],X[:,1],c=y,cmap='winter',s=100)
```

```python
def perceptron(X,y):
  X= np.insert(X,0,1,axis=1)
  weights=np.ones(X.shape[1])
  lr=0.1

  for i in range(1000):             ### looop for 1000 times as outr epoch
    j=np.random.randint(0,100)
    y_hat=step(np.dot(X[j],weights))
    weights=weights + lr*(y[j]-y_hat)*X[j]

  return weights[0],weights[1:]
```

MAking a step function

```python
def step(z):
  return 1 if z>0 else 0        ##   step function used to return output value of the model
```

```python
intercept_,coef_=perceptron(X,y)

print('intercept :',intercept_)
print('coef :',coef_)
```

```python
m = -(coef_[0]/coef_[1])        # finding the values of slope and intercept using the formula m = -a/b , b= -c/b
b = -(intercept_/coef_[1])
```

```python
x_input= np.linspace(-3,3,100)
y_input= m*x_input +b
```

```python
plt.figure(figsize=(10,6))
plt.plot(x_input,y_input,color='red',linewidth=3)
plt.scatter(X[:,0],X[:,1],c=y,cmap='winter',s=-100)
plt.show()
```
