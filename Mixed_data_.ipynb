```python
import pandas as pd
import numpy as np
```

```python
df=pd.read_csv("/content/titanic.csv")
df.head()
```

```python
df["number"].unique()
```

```python
## extract numercal part

df["number_numerical"]=pd.to_numeric(df["number"],errors="coerce",downcast="integer")
```

```python
# extract categorical part
df["number_categorical"]=np.where(df["number_numerical"].isnull(),df["number"],np.nan)

df.head()
```

```python
df["cabin_num"]=df["Cabin"].str.extract("(\d+)")   # captres numerical part
df["cabin_cat"]=df["Cabin"].str[0]    # captures  the first letter

df.head()
```

Working wtih Date and time

```python
# converting  to datetime datatype
date["date"]=pd.to_datetime(date["date"])
```

```python
date.info()
```

Extracting the year

```python
date["date_year"]=date["date"].dt.year    # dt.year extracts the year from the given date
date.head()
```

Extracting the month

```python
date["date_month_no"]=date["date"].dt.month
date.head()
```

```python
## weekend
date["date_dow_name"]=date["date"].dt.day_name()
date["date_is_weekend"]=np.where(date["date_dow_name"].isin(["Sunday","Saturday"]),1,0)
date.drop(columns=["product_id","city_id","orders"]).head()
```

```python
# converting datetime datatype

time["date"]=pd.to_datetime(time["date"])
```

```python
time["hour"]=time["date"].dt.hour
time["min"]=time["date"].dt.minute
time["sec"]=time["date"].dt.second

time.head()
```
