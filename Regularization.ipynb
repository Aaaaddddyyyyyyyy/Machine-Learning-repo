```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
from sklearn.datasets import load_diabetes
data= load_diabetes()
```

```python
print(data.DESCR)
```

```python
X=data.data
y=data.target                   ## extraction of input and output
```

```python
from sklearn.model_selection import train_test_split                  ## train test split
X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=20,random_state=10)
```

```python
from sklearn.linear_model import LinearRegression       # creatiung thwe object
L=LinearRegression()
L.fit(X_train,y_train)        ## training the model
```

```python
y_pred=L.predict(X_test)      ## ppredicting on test dataset using the predict function
```

```python
from sklearn.metrics import r2_score,mean_squared_error
                                                          #$# predicting ther mse and r2 score
print(r2_score(y_test,y_pred))
print(mean_squared_error(y_test,y_pred))
```

```python
from sklearn.linear_model import Ridge        ## training using ridge model
R=Ridge(alpha=0.001)              ## alpha isna hyper paramtrer
R.fit(X_train,y_train)
```

```python
y_pred1=R.predict(X_test)       ## predicting the value of ridge usu9ing prediction functiuon
```

```python
print("r2_score",r2_score(y_test,y_pred1))
print("mse",mean_squared_error(y_test,y_pred1))
```

```python
from sklearn.datasets import  make_regression
import matplotlib.pyplot as plt
import numpy as np
```

```python
X,y=make_regression(n_samples=100,n_features=1, n_informative=1,n_targets=1, noise=10,random_state=34)
```

```python
plt.scatter(X,y)
```

```python
from sklearn.linear_model import LinearRegression
```

```python
lr=LinearRegression()             # makung thew class of linear regression model
lr.fit(X,y)                       ## traing the linear regressiojn moidel
print('coef ',lr.coef_)
print("Intercept",lr.intercept_)
```

```python
from sklearn.linear_model import Ridge

r=Ridge(alpha=10)
r.fit(X,y)
print("coef of ridge",r.coef_)
print("intercept of  ridge",r.intercept_)
```
